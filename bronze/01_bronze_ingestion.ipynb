## Infrastructure Setup

```python
# Infrastructure Setup
spark.sql("CREATE CATALOG IF NOT EXISTS olist")

spark.sql("CREATE SCHEMA IF NOT EXISTS olist.bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS olist.silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS olist.gold")

spark.sql("CREATE VOLUME IF NOT EXISTS olist.bronze.raw_data")
```

## Ingest CSV Files to Delta Tables

### customers 


```python
# Define source path
raw_path = "/Volumes/olist/bronze/raw_data"

# Load and save customers table
df_customers = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{raw_path}/olist_customers_dataset.csv")

df_customers.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("olist.bronze.customers")
```
